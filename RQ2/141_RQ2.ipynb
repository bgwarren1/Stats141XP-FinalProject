```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df = pd.read_csv('/Users/djovanvelasco/Downloads/dat_withSentiment.csv')
df_plot = df.dropna(subset=['ethnicity', 'gender'])
df_plot.head()
sns.set_theme(style="whitegrid", font_scale=1.2)
```

```python

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Subplot A: By Ethnicity
sns.barplot(
    data=df_plot,
    x='community_work_valuable',
    y='ethnicity',
    hue='ethnicity',     # <-- Added to fix the warning
    legend=False,        # <-- Added to fix the warning
    ax=axes[0],
    palette='viridis',
    errorbar='ci',
    capsize=0.1
)
axes[0].set_title('Perceived Value of Community Work\nby Race/Ethnicity', fontsize=16, weight='bold', pad=15)
axes[0].set_xlabel('Average Score (Out of 5)', fontsize=14)
axes[0].set_ylabel('')
axes[0].set_xlim(3, 5.2)

# Subplot B: By Gender
sns.barplot(
    data=df_plot[df_plot['gender'] != 'Prefer not to Say'],
    x='community_work_valuable',
    y='gender',
    hue='gender',        # <-- Added to fix the warning
    legend=False,        # <-- Added to fix the warning
    ax=axes[1],
    palette='viridis',
    errorbar='ci',
    capsize=0.1
)
axes[1].set_title('Perceived Value of Community Work\nby Gender', fontsize=16, weight='bold', pad=15)
axes[1].set_xlabel('Average Score (Out of 5)', fontsize=14)
axes[1].set_ylabel('')
axes[1].set_xlim(3, 5.2)

plt.tight_layout()
plt.savefig('community_work_valuable_demographics.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

# Subplot A: Sentiment by Ethnicity
sns.boxplot(
    data=df_plot,
    x='overall_scenario_sentiment',
    y='ethnicity',
    ax=axes[0],
    color='white', linewidth=2, fliersize=0, boxprops=dict(alpha=0.5)
)
sns.swarmplot(
    data=df_plot,
    x='overall_scenario_sentiment',
    y='ethnicity',
    ax=axes[0],
    palette='viridis', marker='h', size=8, hue='ethnicity', legend=False
)
axes[0].set_title('Reasoning Sentiment Across Scenarios\nby Race/Ethnicity', fontsize=16, weight='bold', pad=15)
axes[0].set_xlabel('Average Sentiment Score (-1 to 1)', fontsize=14)
axes[0].set_ylabel('')

# Subplot B: Sentiment by Gender
sns.boxplot(
    data=df_plot[df_plot['gender'] != 'Prefer not to Say'],
    x='overall_scenario_sentiment',
    y='gender',
    ax=axes[1],
    color='white', linewidth=2, fliersize=0, boxprops=dict(alpha=0.5)
)
sns.swarmplot(
    data=df_plot[df_plot['gender'] != 'Prefer not to Say'],
    x='overall_scenario_sentiment',
    y='gender',
    ax=axes[1],
    palette='viridis', marker='h', size=8, hue='gender', legend=False
)
axes[1].set_title('Reasoning Sentiment Across Scenarios\nby Gender', fontsize=16, weight='bold', pad=15)
axes[1].set_xlabel('Average Sentiment Score (-1 to 1)', fontsize=14)
axes[1].set_ylabel('')

plt.tight_layout()
plt.savefig('scenario_sentiment_demographics.png', dpi=300, bbox_inches='tight')
plt.show() # Displays the plot in the notebook
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# 1. Setup and Melt the Data
df = pd.read_csv('/Users/djovanvelasco/Downloads/dat_withSentiment.csv')
df_plot = df.dropna(subset=['ethnicity', 'gender'])

sentiment_cols = [
    'scenario_1_reason_sentiment', 'scenario_2_reason_sentiment', 
    'scenario_3_reason_sentiment', 'scenario_4_reason_sentiment', 
    'scenario_5_reason_sentiment'
]

# Melting the data transforms it so we can group by Scenario
df_long = pd.melt(
    df_plot, 
    id_vars=['ethnicity', 'gender'], 
    value_vars=sentiment_cols,
    var_name='Scenario', 
    value_name='Sentiment_Score'
)

# Clean up scenario names so they look pretty on the axis
df_long['Scenario'] = df_long['Scenario'].str.replace('_reason_sentiment', '').str.replace('_', ' ').str.title()

# 2. Plotting
sns.set_theme(style="whitegrid", font_scale=1.2)
fig, axes = plt.subplots(2, 1, figsize=(16, 12)) # Taller figure to fit the breakdown

# Subplot A: By Ethnicity
sns.barplot(
    data=df_long, 
    x='Scenario', 
    y='Sentiment_Score', 
    hue='ethnicity', 
    ax=axes[0], 
    palette='viridis',
    errorbar='ci', capsize=0.05
)
axes[0].set_title('Average Reasoning Sentiment per Scenario by Race/Ethnicity', fontsize=16, weight='bold', pad=15)
axes[0].set_ylabel('Average Sentiment Score (-1 to 1)', fontsize=14)
axes[0].set_xlabel('')
axes[0].legend(title='Race/Ethnicity', bbox_to_anchor=(1.02, 1), loc='upper left')

# Subplot B: By Gender
sns.barplot(
    data=df_long[df_long['gender'] != 'Prefer not to Say'], 
    x='Scenario', 
    y='Sentiment_Score', 
    hue='gender', 
    ax=axes[1], 
    palette='viridis',
    errorbar='ci', capsize=0.05
)
axes[1].set_title('Average Reasoning Sentiment per Scenario by Gender', fontsize=16, weight='bold', pad=15)
axes[1].set_ylabel('Average Sentiment Score (-1 to 1)', fontsize=14)
axes[1].set_xlabel('')
axes[1].legend(title='Gender', bbox_to_anchor=(1.02, 1), loc='upper left')

plt.tight_layout()

# SAVE FIRST, THEN SHOW
plt.savefig('scenario_sentiment_breakdown.png', dpi=300, bbox_inches='tight')
plt.show()
```
